```python
class customer:
    def __init__(self, customerid, customername, occupation, age):
        self.customerid = customerid
        self.customername = customername
        self.occupation = occupation
        self.age = age

class policy:
    def __init__(self, policyno, policytype, premiumamount, policyduration, customerid):
        self.policyno = policyno
        self.policytype = policytype
        self.premiumamount = premiumamount
        self.policyduration = policyduration
        self.customerid = customerid 

class agent:
    def __init__(self, agentid, agentname, branchlocation):
        self.agentid = agentid
        self.agentname = agentname
        self.branchlocation = branchlocation

class insurancesystem:
    def __init__(self):
        self.customers = {} 
        self.policies = {}  
        self.agents = {}    

    def add_customer(self):
        try:
            customerid = int(input("Enter customer ID: "))
            if customerid in self.customers:
                raise RuntimeError("The customer ID already exists.")
            customername = input("Enter the customer name: ")
            occupation = input("Enter the occupation: ")
            age = int(input("Enter the age: "))

            new_customer = customer(customerid, customername, occupation, age)
            self.customers[customerid] = new_customer
            print("Successfully added customer details.")
        except RuntimeError as e:
            print(e)
        except ValueError:
            print("Invalid input. Customer ID and age must be numeric.")
        except Exception as e:
            print(f"Error: {e}")

    def add_policy(self):
        try:
            customerid = int(input("Enter customer ID for the policy: "))
            if customerid not in self.customers:
                raise RuntimeError("Customer ID does not exist. Please add customer first.")

            policyno = input("Enter the policy number: ")
            if policyno in self.policies:
                raise RuntimeError("The policy number already exists.")

            policytype = input("Enter the policy type (e.g., health, life, vehicle, property): ")
            premiumamount = int(input("Enter the premium amount: "))
            policyduration = input("Enter the policy duration: ")

            new_policy = policy(policyno, policytype, premiumamount, policyduration, customerid)
            self.policies[policyno] = new_policy
            print("Successfully added policy details.")
        except RuntimeError as e:
            print(e)
        except ValueError:
            print("Invalid input. Customer ID and premium amount must be numeric.")
        except Exception as e:
            print(f"Error: {e}")

    def update_policy(self):
        try:
            policyno = input("Enter the policy number to update: ")
            if policyno not in self.policies:
                raise RuntimeError("The policy number does not exist.")

            current_policy = self.policies[policyno]

            print("Enter new policy details (leave blank to keep current value):")
            new_policytype = input(f"Current Policy Type ({current_policy.policytype}): ")
            if new_policytype:
                current_policy.policytype = new_policytype

            new_premiumamount_str = input(f"Current Premium Amount ({current_policy.premiumamount}): ")
            if new_premiumamount_str:
                current_policy.premiumamount = int(new_premiumamount_str)

            new_policyduration = input(f"Current Policy Duration ({current_policy.policyduration}): ")
            if new_policyduration:
                current_policy.policyduration = new_policyduration

            print("Successfully updated the policy record.")
        except ValueError:
            print("Invalid input. Premium amount must be numeric.")
        except RuntimeError as e:
            print(e)
        except Exception as e:
            print(f"Error: {e}")

    def search_policy(self):
        try:
            policyno = input("Enter the policy number to search: ")
            if policyno not in self.policies:
                raise KeyError("The policy number does not exist.")

            p = self.policies[policyno]
            c = self.customers.get(p.customerid) # Get customer details

            print("\nPolicy Details")
            if c:
                print(f"Customer ID: {c.customerid}")
                print(f"Customer Name: {c.customername}")
                print(f"Occupation: {c.occupation}")
                print(f"Age: {c.age}")
            else:
                print(f"Customer details not found for ID: {p.customerid}")

            print(f"Policy Number: {p.policyno}")
            print(f"Policy Type: {p.policytype}")
            print(f"Premium Amount: {p.premiumamount}")
            print(f"Policy Duration: {p.policyduration}")
        except KeyError as e:
            print(e)
        except Exception as e:
            print(f"Error: {e}")

    def display_policy(self): 
        if not self.policies:
            print("No policy records available.")
            return

        print("\nAll Policy Details")

        for p in self.policies.values():
            c = self.customers.get(p.customerid) # Get customer details
            print("-----------------------------")
            if c:
                print(f"Customer ID: {c.customerid}")
                print(f"Customer Name: {c.customername}")
                print(f"Occupation: {c.occupation}")
                print(f"Age: {c.age}")
            else:
                print(f"Customer details not found for ID: {p.customerid}")

            print(f"Policy Number: {p.policyno}")
            print(f"Policy Type: {p.policytype}")
            print(f"Premium Amount: {p.premiumamount}")
            print(f"Policy Duration: {p.policyduration}")

    def add_agent(self):
        try:
            agentid = int(input("Enter agent ID: "))
            if agentid in self.agents:
                raise RuntimeError("The agent ID already exists.")
            agentname = input("Enter the agent name: ")
            branchlocation = input("Enter the branch location: ")
            new_agent = agent(agentid, agentname, branchlocation)
            self.agents[agentid] = new_agent
            print("Successfully added agent details.")
        except RuntimeError as e:
            print(e)
        except ValueError:
            print("Invalid input. Agent ID must be numeric.")
        except Exception as e:
            print(f"Error: {e}")

    def display_agent(self):
        if not self.agents:
            print("No agent records available.")
            return

        print("\nAll Agent Details")

        for a in self.agents.values():
            print("-----------------------------")
            print(f"Agent ID: {a.agentid}")
            print(f"Agent Name: {a.agentname}")
            print(f"Branch Location: {a.branchlocation}")

    def menu(self): 
        while True:
            print("\nInsurance Policy Management System")
            print("1. Add customer")
            print("2. Add policy")
            print("3. Update policy")
            print("4. Search policy")
            print("5. Display All Policies")
            print("6. Add agent")
            print("7. Display All agents")
            print("8. Exit")

            try:
                choice = int(input("Enter choice: "))

                if choice == 1:
                    self.add_customer()
                elif choice == 2:
                    self.add_policy()
                elif choice == 3:
                    self.update_policy()
                elif choice == 4:
                    self.search_policy()
                elif choice == 5:
                    self.display_policy() 
                elif choice == 6:
                    self.add_agent()
                elif choice == 7:
                    self.display_agent()
                elif choice == 8:
                    print("Exiting system...")
                    break
                else:
                    print("Invalid choice. Please enter a number from 1 to 8.")

            except ValueError:
                print("Invalid input. Please enter a numeric choice only.")
            except Exception as e:
                print(f"An unexpected error occurred: {e}")

system = insurancesystem()
system.menu()
```
